```python
def clean_sheet(sheet, sector, has_program):
    # Read the sheet (column names are on the 2nd row)
    df = pd.read_excel(file, sheet_name=sheet, header=1)

    # Keep only the real columns (some sheets have thousands of empty ones)
    n_cols = 5 if has_program else 4
    df = df.iloc[:, :n_cols]

    # Short column names
    if has_program:
        df.columns = ["institution", "program", "loan_recipients", "rap_users", "rap_rate"]
    else:
        df.columns = ["institution", "loan_recipients", "rap_users", "rap_rate"]

    # Trim extra spaces from names
    df["institution"] = df["institution"].astype("string").str.strip()

    # Keep only rows above "Notes:" (if the sheet has notes)
    notes = df.index[df["institution"] == "Notes:"]
    if len(notes) > 0:
        df = df.loc[:notes[0] - 1]

    # Remove junk rows: blanks, "*" rows, and any TOTAL rows
    df = df.dropna(subset=["institution"])
    df = df[df["institution"] != "*"]
    df = df[~df["institution"].str.upper().str.contains("TOTAL")]

    # Trim spaces from program names too
    if has_program:
        df["program"] = df["program"].astype("string").str.strip()

    # Flag hidden (privacy) values, then turn "*" into blanks
    df["suppressed"] = df["loan_recipients"] == "*"
    for col in ["loan_recipients", "rap_users", "rap_rate"]:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    # Add sector and year
    df["sector"] = sector
    df["year"] = 2020
    return df
```

```python
institution_sheets = {"University": "University", "College": "College",
                      "PCC": "Private Career College", " Other": "Other"}
program_sheets = {"University Program": "University", "College Program": "College",
                  "PCC Program": "Private Career College", "Other by Program": "Other"}

inst = pd.concat([clean_sheet(s, sec, False) for s, sec in institution_sheets.items()],
                 ignore_index=True)
prog = pd.concat([clean_sheet(s, sec, True) for s, sec in program_sheets.items()],
                 ignore_index=True)

print(inst.shape, prog.shape)
```

```python
inst.to_csv("osap_rap_2020_institutions_clean.csv", index=False)
prog.to_csv("osap_rap_2020_programs_clean.csv", index=False)
```

```python
## Data cleaning log
- Removed title rows, "*" header rows, TOTAL/subtotal rows, blank rows and footnotes.
- Trimmed extra spaces from institution and program names.
- "*" values = suppressed for privacy (small counts). Kept the rows, set numbers to blank, flagged suppressed = True.
- Validated sector totals against the official Sector Summary. College matches exactly; other gaps are due to suppressed values.
- Duplicates found: "MASTERS COLLEGE AND SEMINARY" (2 rows, one suppressed) and "LA CITE COLLEGIALE - Parajuriste" (2 rows, both suppressed). Kept both; no effect on totals.
- Source data issue: some sheets in the 2020 file are labelled "2019 RAP Usage Rate".
```
